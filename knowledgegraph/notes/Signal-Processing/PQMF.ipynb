```python
import numpy as np
import matplotlib.pyplot as plt

N = 2048

def plot_spectrum(x, real=False, label=None, alpha=1.0):
    if real:
        spec = np.fft.rfft(x)
    else:
        spec = np.fft.fft(x)
    plt.plot(np.abs(spec), label=label, alpha=alpha)
    plt.ylabel("Amplitude")
    plt.xlabel("Frequency (DFT Index)")
    plt.legend()
    plt.ylim(-0.1,1.1)


def filter(x, lowpass=True):
    spec = np.fft.rfft(x)
    cutoff = spec.shape[0] // 2

    if lowpass:
        spec[cutoff:] = 0
    else:
        spec[:cutoff] = 0

    return np.fft.irfft(spec)

def stretch(x):
    zeros = np.zeros(x.shape[-1])
    return np.stack((x, zeros)).T.reshape(-1)

real_spectrum = np.flip(np.arange(N)) / N
plt.plot(np.abs(real_spectrum))
x = np.fft.irfft(real_spectrum)
plot_spectrum(x, real=True)
```

```python
plot_spectrum(x, label="Original Spectrum")
plt.title("Original Spectrum")
plt.savefig("PQMF-1-Original-Signal.svg")
```

```python
x_low = filter(x, lowpass=True)
x_high = filter(x, lowpass=False)
plot_spectrum(x_low, label="Low-Passed Signal", alpha=0.9)
plot_spectrum(x_high, label="High-Passed Signal", alpha=0.9)
plt.title("Comparing Low and High Band of the Signal")
plt.savefig("PQMF-2-Low-High-Compare.svg")
```

```python
x_low_downsampled = 2*x_low[::2]
x_high_downsampled = 2*x_high[::2]
plot_spectrum(x_low, label="Low-Passed Signal")
plot_spectrum(x_low_downsampled, label="Low-Passed Signal, After Downsampling", alpha=0.7)
plt.title("Comparing Low-Passed Signal Before and After Downsampling")
plt.savefig("PQMF-3-Low-Downsample-Compare.svg")
plt.show()
plot_spectrum(x_high, label="High-Passed Signal")
plot_spectrum(x_high_downsampled, label="High-Passed Signal, After Downsampling",alpha=0.7)
plt.title("Comparing High-Passed Signal Before and After Downsampling")
plt.savefig("PQMF-4-High-Downsample-Compare.svg")
plt.show()
```

```python
x_low_downsampled_stretched = stretch(x_low_downsampled)
x_high_downsampled_stretched = stretch(x_high_downsampled)

plot_spectrum(x_low_downsampled, label="Low-Passed Signal, After Downsampling", alpha=1.0)
plot_spectrum(
    x_low_downsampled_stretched,
    label="Low-Passed Signal, After Downsampling and Stretching",
    alpha=0.5
)
plt.savefig("PQMF-5-Low-Stretch-Compare.svg")
plt.show()



plt.show()
plot_spectrum(x_high_downsampled, label="High-Passed Signal, After Downsampling", alpha=1.0)
plot_spectrum(x_high_downsampled_stretched, label="High-Passed Signal, After Downsampling and Stretching", alpha=0.5)
plt.savefig("PQMF-6-High-Stretch-Compare.svg")
plt.show()
```

```python
x_low_downsampled_stretched_low_passed = filter(x_low_downsampled_stretched, lowpass=True)
x_high_downsampled_stretched_high_passed = filter(x_high_downsampled_stretched, lowpass=False)

plot_spectrum(
    x_low_downsampled_stretched_low_passed,
    label="Low-Passed Signal, After Downsampling, Stretching and Low-Passing Again",
    alpha=0.7,
)
plot_spectrum(
    x_high_downsampled_stretched_high_passed,
    label="High-Passed Signal, After Downsampling, Stretching, and High-Passing Again",
    alpha=0.7,
)
plt.savefig("PQMF-7-Reconstruction.svg")
plt.show()
```
